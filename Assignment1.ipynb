<h2>Preprocessing</h2>


<h3>Load Data</h3>

<h3>Cleaning Raw Data</h3>

```python
# Stopwords are commonly occurring words in a language like ‘the’, ‘a’, and so on.
# They can be removed from the text most of the time, as they don’t provide valuable information.
from nltk.corpus import stopwords
stopwords = stopwords.words('english')

def preprocessing(text: pd.Series):
    clean_text =[]
    # text.str.lower = Lowering each word of text in all columns
    for sentence in text.str.lower():
        sentence = (re.sub("\.", "", sentence))
        # replaces special symbols @ # $ §, etc. with " "
        sentence = (re.sub('[^A-Za-z0-9]+', ' ', sentence))
        
        # Regular expression pattern for dates in the format yyyy-mm-dd
        date_pattern1 = r'\d{4}-\d{2}-\d{2}'
        date_pattern2 = r'\d{2}-\d{2}-\d{4}'
        # Regular expression pattern for numbers with optional decimal points
        number_pattern = r'\d+'
        sentence = (re.sub(date_pattern1, "<dates>", sentence))
        sentence = (re.sub(date_pattern2, "<dates>", sentence))
        sentence = (re.sub(number_pattern, "<num>", sentence))
        
        # Words like ‘the’, ‘to’, ‘at’ are removed
        sentence = ' '.join(e.lower() for e in sentence.split() if e.lower() not in stopwords)
        
        #Removal any URLs and HTML tags
        sentence = re.sub(r"http\S+", "", sentence)
        sentence = re.sub(r"'<.*?>'", "", sentence)
    
        clean_text.append(sentence)
    return clean_text
```

```python
print(f"Before Cleaning Raw Data (aka. Preprocessing):\n {load_train_text.Text[0]}")
pre_train_text = preprocessing(load_train_text.Text)
pre_train_text = pd.Series(pre_train_text)

pre_validation_text = preprocessing(load_validation_text.Text)
pre_validation_text = pd.Series(pre_validation_text)

pre_test_text= preprocessing(load_test_text.Text)
pre_test_text = pd.Series(pre_test_text)
print(f"After Cleaning Raw Data (aka. Preprocessing):\n {pre_train_text[0]}")
```

<h3>Tokenization</h3>

<h4>Porter’s stemmer & Lemmatization functions </h4>

```python
from nltk.stem.porter import PorterStemmer
from nltk.stem import WordNetLemmatizer

# Porter’s stemmer 
# A stemmer reduces words to their “stems” by crude affix chopping
# automate, automates, automatic, automation -> automat
def stem_words(list_of_tokens: []):
    ps = PorterStemmer()
    
    stem_words = []
    for word in list_of_tokens:
        if word != "<num>" or word != "<dates>":
            stem_words.append(ps.stem(word))
        else:
            stem_words.append(word)
    return stem_words

# Lemmatization
# lemmatizer uses a knowledge resource (like WordNet) to find and replace base forms
# stemming ist faster, but also harms precision and increase ambiguity, therefore we choose Lemmatization
def lem_words(list_of_tokens: []):
    lemmatizer = WordNetLemmatizer()
    lem_words = []
    for word in list_of_tokens:
        if word != "<num>" or word != "<dates>":
            lem_words.append(lemmatizer.lemmatize(word))
        else:
            lem_words.append(word)
            
    return lem_words
```

<h4>Tokenize data</h4>

```python
def tokenization(text: pd.Series, normalization_method: str = "lem"):
    corpus = []
    for sentence in text:
        # not using word_tokenize, because i want to have "<num>" and not "<","num", ">"
        corpus.append(sentence.split())
        
    list_of_words = np.concatenate(corpus)
    # gets normalized with one of the methods
    # default method is lem_words
    if normalization_method == "lem":
        return lem_words(list_of_words), corpus
    else:
        return stem_words(list_of_words), corpus
```

```python
train_tokens, train_corpus = tokenization(pre_train_text)
validation_tokens, validation_corpus = tokenization(pre_validation_text)
test_tokens, test_corpus = tokenization(pre_test_text)
print(f"50 Tokens from train_tokens:\n {train_tokens[0:49]}")
```

<h2>Creating dictionary</h2>

```python
# Counter counts all words + each frequency
dictionary_of_all_words = Counter(train_tokens)

# threshold is 80 % of bag of words
n = 0.4
threshold = int(len(dictionary_of_all_words) * n)
dictionary_with_threshold = heapq.nlargest(threshold, dictionary_of_all_words, key=dictionary_of_all_words.get)

# creating list of unique tokens of train_tokens
used = set()
unique_train_tokens = [x for x in train_tokens if x not in used and (used.add(x) or True)]

# OOV
out_of_vocabulary = list(set(unique_train_tokens) - set(dictionary_with_threshold))

print(f"Length of dictionary without threshold: {len(dictionary_of_all_words)}")
print(f"Length of dictionary with threshold of {n * 100}%: {len(dictionary_with_threshold ) -1 } + 1 = {len(dictionary_with_threshold)} (+ 1 so we have <oov> also in the dictionary) ")
print(f"Length of the OOVs: {len(out_of_vocabulary)}")
```

<h3>Handeling OOVs</h3>

<h4>Replacing OOV-words in train/val/test</h4>

```python
def replacing_OOV(text: pd.Series, oovs: []):
    new_sentence = []
    for idx, sentence in enumerate(tqdm(text, desc="Detecting OOVs")):
        # make regex pattern with the OVV words in sentence
        oov_pattern = re.compile(r'\b(?:' + '|'.join(map(re.escape, oovs)) + r')\b', re.IGNORECASE)
        # replace OOV word with "OOV"
        new_sentence.append(re.sub(oov_pattern, "<oov>", sentence))
    return pd.Series(new_sentence)
```

```python
train_text_with_oov = replacing_OOV(text=pre_train_text, oovs=out_of_vocabulary)
validation_text_with_oov = replacing_OOV(text=pre_validation_text, oovs=out_of_vocabulary)
test_text_with_oov = replacing_OOV(text=pre_test_text, oovs=out_of_vocabulary)
```

<h3>Setup Doc-Word-Matrices for Train / Val / Test</h3>

```python
# helper function for df
def get_df(word:str, corpus: list):
    word_count = 0
    for sentence in corpus:
        if word in sentence:
            word_count += 1
        else:
            continue
    return word_count
```

<h4>TF-IDF Method</h4>

```python
def tf_idf(train_text, dictonary, corpus):
    # creating Document-Word-matrix
    doc_word_matrix = np.zeros(shape=(len(train_text), len(dictonary)))
    
    # amount of documents 
    M = doc_word_matrix.shape[0]
    
    for idx_d_row, d in enumerate(tqdm(doc_word_matrix, desc="Processing Documents")):
        # list of word in document
        document = np.array(train_text[idx_d_row].split())
        for word in document:
            # get index position of word in dictonary
            idx = np.where(np.asarray(dictonary) == word)[0]
            
            # calc term frequency tf
            # tc is the number occurrences of word in document
            tc_vd = len(np.where(document == word)[0])
            tf_vd = np.log(1 + tc_vd)
            
            # calc inverse document frequency idf_v
            # document frequency is the number of doc that contain the word
            df_v = get_df(word=word, corpus=corpus)
            # high number means word appears less often and is therefore more informative
            idf_v = np.log(M/(df_v + 1)) 
            
            tf_idf = tf_vd * idf_v
            
            doc_word_matrix[idx_d_row][idx] = np.round(tf_idf, 5)
    return doc_word_matrix
```

<h4>BM25 Method</h4>

```python
def bm25(train_text, dictonary, corpus, k1: float = 0.6, b: float = 1.0):
    # creating Document-Word-matrix
    doc_word_matrix = np.zeros(shape=(len(train_text), len(dictonary)))
    # amount of documents 
    M = doc_word_matrix.shape[0]
    
    # average lenght of documents in collection
    avgdl = 0
    for sentences in corpus:
        avgdl += len(sentences)
    avgdl = avgdl / M
    
    
    for idx_d_row, docs in enumerate(tqdm(doc_word_matrix, desc="Processing Documents")):
        # list of word in document
        document = np.array(train_text[idx_d_row].split())
        d = len(document)
        for word in document:
            # get index position of word in dictonary
            idx = np.where(np.asarray(dictonary) == word)[0]
            
            # calc (k1 + 1) * tc_vd
            # tc is the number occurrences of word in document
            tc_vd = len(np.where(document == word)[0])
            top_part_of_bm25 = (k1 + 1) * tc_vd
            
            # calc (k1 * (1-b+b * (|d|/avgdl)) + tc_vd)
            down_part_of_bm25 = (k1 * (1 - b + b * (d/avgdl)) ) + tc_vd
            
            # calc inverse document frequency idf_v
            # document frequency is the number of doc that contain the word
            df_v = get_df(word=word, corpus=corpus)
            # high number means word appears less often and is therefore more informative
            idf_v = np.log(M/(df_v + 1)) 
            
            bm25_vd = (top_part_of_bm25 / down_part_of_bm25) * idf_v
            
            doc_word_matrix[idx_d_row][idx] = np.round(bm25_vd, 5)
    return doc_word_matrix
```
