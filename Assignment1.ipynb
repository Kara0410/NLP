<h2>Used Libraries</h2>

```python
import pandas as pd
import re
import nltk
from collections import Counter
import heapq
import numpy as np
import itertools

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')
```

<h2>Preprocessing</h2>

<h3>Load Data</h3>

```python
train_text = pd.read_csv("nlp2023_24_data/thedeep.subset.train.txt", header=None, names=["Number", "Text", "Label"])
validation_text = pd.read_csv("nlp2023_24_data/thedeep.subset.validation.txt", header=None, names=["Number", "Text", "Label"])
test_text = pd.read_csv("nlp2023_24_data/thedeep.subset.test.txt", header=None, names=["Number", "Text", "Label"])
train_text.head()
```

<h3>Cleaning Raw Data</h3>

```python
# Stopwords are commonly occurring words in a language like ‘the’, ‘a’, and so on.
# They can be removed from the text most of the time, as they don’t provide valuable information.
from nltk.corpus import stopwords
stopwords = stopwords.words('english')

print(f"Before Cleaning Raw Data (aka. Preprocessing):\n {train_text.Text[0]}")
def preprocessing(text: pd.Series):
    clean_text =[]
    # text.str.lower = Lowering each word of text in all columns
    for sentence in text.str.lower():
        # for decimal numbers
        sentence = (re.sub('\.', '', sentence))
        
        # replaces special symbols @ # $ §, etc. with " "
        sentence = (re.sub('[^A-Za-z0-9]+', ' ', sentence))
        
        # Regular expression pattern for dates in the format yyyy-mm-dd
        date_pattern1 = r'\d{4}-\d{2}-\d{2}'
        date_pattern2 = r'\d{2}-\d{2}-\d{4}'
        # Regular expression pattern for numbers with optional decimal points
        number_pattern = r'\d+'
        sentence = (re.sub(date_pattern1, "<dates>", sentence))
        sentence = (re.sub(date_pattern2, "<dates>", sentence))
        sentence = (re.sub(number_pattern, "<num>", sentence))
        
        # Words like ‘the’, ‘to’, ‘at’ are removed
        sentence = ' '.join(e.lower() for e in sentence.split() if e.lower() not in stopwords)
        
        #Removal any URLs and HTML tags
        sentence = re.sub(r"http\S+", "", sentence)
        sentence = re.sub(r"'<.*?>'", "", sentence)
    
        clean_text.append(sentence)
    return clean_text
```

```python
train_text.Text = preprocessing(train_text.Text)
validation_text.Text = preprocessing(validation_text.Text)
test_text.Text = preprocessing(test_text.Text)
print(f"After Cleaning Raw Data (aka. Preprocessing):\n {train_text.Text[0]}")
```

```python
train_text.Text
```

<h3>Tokenization</h3>

<h4>Porter’s stemmer & Lemmatization functions </h4>

```python
from nltk.stem.porter import PorterStemmer
from nltk.stem import WordNetLemmatizer

# Porter’s stemmer 
# A stemmer reduces words to their “stems” by crude affix chopping
# automate, automates, automatic, automation -> automat
def stem_words(list_of_tokens: []):
    ps = PorterStemmer()
    
    stem_words = []
    for word in list_of_tokens:
        if word != "<num>" or word != "<dates>":
            stem_words.append(ps.stem(word))
        else:
            stem_words.append(word)
    return stem_words

# Lemmatization
# lemmatizer uses a knowledge resource (like WordNet) to find and replace base forms
# stemming ist faster, but also harms precision and increase ambiguity, therefore we choose Lemmatization
def lem_words(list_of_tokens: []):
    lemmatizer = WordNetLemmatizer()
    lem_words = []
    for word in list_of_tokens:
        if word != "<num>" or word != "<dates>":
            lem_words.append(lemmatizer.lemmatize(word))
        else:
            lem_words.append(word)
            
    return lem_words
```

<h4>Tokenize data</h4>

```python
def tokenization(text: pd.Series, normalization_method: str = "lem"):
    list_of_words = []
    for sentence in text:
        # not using word_tokenize, because i want to have "<num>" and not "<","num", ">"
        list_of_words.append(sentence.split())
        
    # gets normalized with one of the methods
    # default method is lem_words
    if normalization_method == "lem":
        return lem_words(np.concatenate(list_of_words)), list_of_words
    else:
        return stem_words(np.concatenate(list_of_words)), list_of_words
```

```python
train_tokens, train_corpus = tokenization(train_text.Text)
print(f"50 tokens of training_text:\n{train_tokens[0:49]}")
```

<h2>Creating dictionary</h2>

```python
# Counter counts all words + each frequency
dictionary_of_all_words = Counter(train_tokens)
print(f"Length of dictionary without threshold: {len(dictionary_of_all_words)}")

# threshold is 40 % of bag of words
n = 0.35
threshold = int(len(dictionary_of_all_words) * n)
dictionary_with_threshold = heapq.nlargest(threshold, dictionary_of_all_words, key=dictionary_of_all_words.get)
dictionary_with_threshold.append("<oov>")
print(f"Length of dictionary with threshold of {n * 100.0}% : {len(dictionary_with_threshold) - 1 } + 1 = {len(dictionary_with_threshold)}")
print("+ 1 because we also need to add the token for OOVs <oov> to the dictionary")

# creating list of unique tokens of train_tokens
used = set()
unique_train_tokens = [x for x in train_tokens if x not in used and (used.add(x) or True)]

# OOV
out_of_vocabulary = list(set(unique_train_tokens) - set(dictionary_with_threshold))
print(f"Length of Out of Vocabulary (OOV): {len(out_of_vocabulary)}")
```

<h3>Handeling OOVs</h3>

<h4>Replacing OOV-words in train/val/test</h4>

```python
def replacing_OOV(text: pd.Series, oovs: []):
    senteces_without_oov = []
    for sentence in text:
        sentence = np.asarray(sentence.split())

        # get indeces of OOV words in looped sentence
        oov_words_position_in_sentence = np.where(np.isin(sentence, oovs))[0]

        # if no OOV words are included in sentece skip sentence
        if len(oov_words_position_in_sentence) == 0:
            # sentence to string again
            sentence = " ".join(sentence)
            senteces_without_oov.append(sentence)
        else:
            # make regex pattern with the OVV words in sentence
            pattern = r"|".join(sentence[oov_words_position_in_sentence])
            # sentence to string again
            sentence = " ".join(sentence)
            # replace OOV word with "OOV"
            sentence = (re.sub(pattern, "<oov>", sentence))
            senteces_without_oov.append(sentence)
    return senteces_without_oov
```

```python
train_text.Text = replacing_OOV(text=train_text.Text, oovs=out_of_vocabulary)
train_text.Text
```

<h2>Creating sentence vectors</h2>

```python
# creating Document-word matrices
doc_word_matrix_train = np.zeros(shape=(len(train_text), len(dictionary_with_threshold)))
doc_word_matrix_valid = np.zeros(shape=(len(validation_text), len(dictionary_with_threshold)))
doc_word_matrix_test = np.zeros(shape=(len(test_text), len(dictionary_with_threshold)))
```

```python
# df is the number of documents that contain word 𝑣
def df(word: str, corpus: []):
    df_count = 0
    for doc in corpus:
        if word in doc:
            df_count += 1
        else:
            continue
    return df_count
```

```python
for idx_d_row, d in enumerate(doc_word_matrix_train):
    for idx_v_col, v in enumerate(d):
        
        # gets the sentence from train/val/test set
        sentence = np.asarray(train_text.Text[idx_d_row].split())
        # gets the word from dictionary
        word = dictionary_with_threshold[idx_v_col]
        
        # occurrence of word 𝑣 in 𝑑
        tc_v_d = len(np.where(sentence == word)[0])
        # Term frequency 
        tf_v_d = np.log(1 + tc_v_d)
        
        # document frequency of 𝑣
        df_v = df(word=word, corpus=train_corpus)
        # Inverse document frequency 
        idf_v = np.log(len(train_text)/(df_v + 1))
        
        tf_idf = tf_v_d * idf_v
        
        doc_word_matrix_train[idx_d_row][idx_v_col] = tf_idf
```

```python
doc_word_matrix_train
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Create a CountVectorizer for Term Count (TC)
count_vectorizer = CountVectorizer()
count_train_vectors = count_vectorizer.fit_transform(train_text.Text)
count_validation_vectors = count_vectorizer.transform(validation_text.Text)
count_test_vectors = count_vectorizer.transform(test_text.Text)

# Create a TfidfVectorizer for TF-IDF
tfidf_vectorizer = TfidfVectorizer()
tfidf_train_vectors = tfidf_vectorizer.fit_transform(train_text.Text)
tfidf_validation_vectors = tfidf_vectorizer.transform(validation_text.Text)
tfidf_test_vectors = tfidf_vectorizer.transform(test_text.Text)
```

```python
# Calculate sparsity rate for each set
def calculate_sparsity_rate(vector):
    total_elements = vector.shape[0] * vector.shape[1]
    zero_elements = total_elements - vector.count_nonzero()
    sparsity_rate = (zero_elements / total_elements) * 100
    return sparsity_rate

sparsity_rate_train_count = calculate_sparsity_rate(count_train_vectors)
sparsity_rate_validation_count = calculate_sparsity_rate(count_validation_vectors)
sparsity_rate_test_count = calculate_sparsity_rate(count_test_vectors)

sparsity_rate_train_tfidf = calculate_sparsity_rate(tfidf_train_vectors)
sparsity_rate_validation_tfidf = calculate_sparsity_rate(tfidf_validation_vectors)
sparsity_rate_test_tfidf = calculate_sparsity_rate(tfidf_test_vectors)

# Report the approaches and sparsity rates
print("Term Count (TC) Approach:")
print(f"Train Set Sparsity Rate: {sparsity_rate_train_count:.2f}%")
print(f"Validation Set Sparsity Rate: {sparsity_rate_validation_count:.2f}%")
print(f"Test Set Sparsity Rate: {sparsity_rate_test_count:.2f}%")

print("\nTF-IDF Approach:")
print(f"Train Set Sparsity Rate: {sparsity_rate_train_tfidf:.2f}%")
print(f"Validation Set Sparsity Rate: {sparsity_rate_validation_tfidf:.2f}%")
print(f"Test Set Sparsity Rate: {sparsity_rate_test_tfidf:.2f}%")
```
