```python
import pandas as pd
import re
import nltk
from collections import Counter
import heapq
import numpy as np
import itertools

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')
```

```python
train_text = pd.read_csv("nlp2023_24_data/thedeep.subset.train.txt", header=None, names=["Number", "Text", "Label"])
validation_text = pd.read_csv("nlp2023_24_data/thedeep.subset.validation.txt", header=None, names=["Number", "Text", "Label"])
test_text = pd.read_csv("nlp2023_24_data/thedeep.subset.test.txt", header=None, names=["Number", "Text", "Label"])
train_text.head()
```

```python
# Stopwords are commonly occurring words in a language like ‘the’, ‘a’, and so on.
# They can be removed from the text most of the time, as they don’t provide valuable information.
from nltk.corpus import stopwords
stopwords = stopwords.words('english')

print(f"Before Cleaning Raw Data (aka. Preprocessing):\n {train_text.Text[0]}")
def preprocessing(text: pd.Series):
    clean_text =[]
    # text.str.lower = Lowering each word of text in all columns
    for sentence in text.str.lower():
        # replaces special symbols @ # $ §, etc. with " "
        sentence = (re.sub('[^A-Za-z0-9]+', ' ', sentence))
        
        # Regular expression pattern for dates in the format yyyy-mm-dd
        date_pattern = r'\d{4}-\d{2}-\d{2}'
        # Regular expression pattern for numbers with optional decimal points
        number_pattern = r'\d+(\.\d+)?'
        sentence = (re.sub(date_pattern, "<dates>", sentence))
        sentence = (re.sub(number_pattern, "<num>", sentence))
        
        # Words like ‘the’, ‘to’, ‘at’ are removed
        sentence = ' '.join(e.lower() for e in sentence.split() if e.lower() not in stopwords)
        
        #Removal any URLs and HTML tags
        sentence = re.sub(r"http\S+", "", sentence)
        sentence = re.sub(r"'<.*?>'", "", sentence)
    
        clean_text.append(sentence)
    return clean_text
    
train_text.Text = preprocessing(train_text.Text)
validation_text.Text = preprocessing(validation_text.Text)
test_text.Text = preprocessing(test_text.Text)

print(f"After Cleaning Raw Data (aka. Preprocessing):\n {train_text.Text[0]}")
```

```python
from nltk.stem.porter import PorterStemmer
from nltk.stem import WordNetLemmatizer

# Porter’s stemmer 
# A stemmer reduces words to their “stems” by crude affix chopping
# automate, automates, automatic, automation -> automat
def stem_words(list_of_tokens: []):
    ps = PorterStemmer()
    
    stem_words = []
    for word in list_of_tokens:
        if word != "<num>" or word != "<dates>":
            stem_words.append(ps.stem(word))
        else:
            stem_words.append(word)
    return stem_words

# Lemmatization
# lemmatizer uses a knowledge resource (like WordNet) to find and replace base forms
# stemming ist faster, but also harms precision and increase ambiguity, therefore we choose Lemmatization
def lem_words(list_of_tokens: []):
    lemmatizer = WordNetLemmatizer()
    
    lem_words = []
    for word in list_of_tokens:
        if word != "<num>" or word != "<dates>":
            lem_words.append(lemmatizer.lemmatize(word))
        else:
            lem_words.append(word)
            
    return lem_words

def tokenization(text: pd.Series, normalization_method: str = "stem"):
    list_of_words = []
    
    for sentence in text:
        # not using word_tokenize, because i want to have "<num>" and not "<","num", ">"
        list_of_words += sentence.split()
        
    # gets normalized with one of the methods
    # default method is lem_words
    if normalization_method == "lem":
        return lem_words(list_of_words)
    else:
        return stem_words(list_of_words)

train_tokens = tokenization(train_text.Text)
validation_tokens = tokenization(validation_text.Text)
test_tokens = tokenization(test_text.Text)
```

```python

# Counter counts all words
word2count = Counter(train_tokens)
# sorting dict from highest to lowest
sorted_word2count = {k: v for k, v in sorted(word2count.items(), key=lambda item: item[1], reverse=True)}

# creating list of unique tokens of train_tokens
used = set()
unique_train_tokens = [x for x in train_tokens if x not in used and (used.add(x) or True)]

# 1500 is the threshold that we choose
freq_words = heapq.nlargest(1800, word2count, key=word2count.get)

# OOV
out_of_vocabulary = list(set(unique_train_tokens) - set(freq_words))
```

```python
def document_frequency(list_of_words: [], word: str):
    word_count = 0
    for list_of_word in list_of_words:
        if word in list_of_word:
            word_count += 1
    return word_count
```

```python
# list, which contains each list of word of documents
def corpus_calc (text: pd.Series):
    list_of_words = []
    for sentence in text:
        list_of_words.append(sentence.split())
    return list_of_words



from rank_bm25 import BM25Okapi
corpus = corpus_calc(train_text.Text)
# Create a dictionary and a document-term matrix from your corpus
bm25 = BM25Okapi(corpus)
query = ['addition', 'immediate', 'life', 'saving', 'interventions',
         'unicef', 'taking', 'action', 'protect', '<num>', 'children',
         'arrived', 'camps', 'angola', 'without', 'families']
scores = bm25.get_scores(query)

scores
```

```python
# creating bag of Document-Word matrix
row_len = len(train_text.Number)
col_len = len(freq_words)

corpus = corpus_calc(train_text.Text)
sum_of_tokens = len(list(itertools.chain.from_iterable(corpus)))
avgdl = sum_of_tokens / len(corpus)

dw_matrix = np.zeros((row_len, col_len))

for idx_row, _ in enumerate(dw_matrix):
    document_d = corpus[idx_row]
    number_of_tokens_in_d = len(document_d)
    print(document_d)
    for word in document_d:
        query = []
        if word in freq_words:
            query.append(word)
            # term count slide 1 page 18
            # occurrences of word v in this document d (.aka referring_text)
            tc_vd = Counter(document_d)[word]
            
            # term frequency slide 1 page 19
            tf_vd = np.log(1 + tc_vd)
            
            # the number of documents that contain word v
            df_v = document_frequency(corpus, word=word)
            # Inverse document frequency
            # row_len is the number of documents
            idf_v = np.log(row_len/df_v + 1)
            
            bm25_up_part = (0.6 + 1) * tc_vd
            bm25_down_part = 0.6 * (1 - 1 + 1* ( number_of_tokens_in_d / avgdl) ) * tc_vd
            
            bm25 = (bm25_up_part/bm25_down_part) * idf_v
            
            idx_col = freq_words.index(word)
            
            dw_matrix[idx_row][idx_col] = bm25
        print(query)
        scores = bm25.get_scores(query)
        print(scores)
    print(dw_matrix[0, :])
    break
np.where(dw_matrix[0] != 0 )
```
