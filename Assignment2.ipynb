```python
# Task A.1
import pandas as pd

# path to txt file
glove_file_path = "./glove.6B/glove.6B.50d.txt"

# creating Dataframe from the pre-trained word embedding
word_embedding_model  = pd.read_csv(glove_file_path, sep=" ", header=None, quoting=3)

# shows some tokens from data 
word_embedding_model[500:510]
```

```python
# Task A.2
# Calculating word-to-word similarities
import numpy as np

# cosine similarity function
def cosine_similarity(vector_a, vector_b):
    
    dot_product = np.dot(vector_a, vector_b)
    
    norm_a = np.linalg.norm(vector_a)
    norm_b = np.linalg.norm(vector_b)
    
    similarity = dot_product / (norm_a * norm_b)
    return similarity

# Source words and their vectors
source_words = {
    'dog': word_embedding_model.loc[word_embedding_model[0] == 'dog'].to_numpy()[:, 1:][0],
    'car': word_embedding_model.loc[word_embedding_model[0] == 'car'].to_numpy()[:, 1:][0],
    'computer': word_embedding_model.loc[word_embedding_model[0] == 'computer'].to_numpy()[:, 1:][0],
    'happy': word_embedding_model.loc[word_embedding_model[0] == 'happy'].to_numpy()[:, 1:][0],
    'science': word_embedding_model.loc[word_embedding_model[0] == 'science'].to_numpy()[:, 1:][0]
}

# Target words for each source word
target_words = {
    'dog': ['puppy', 'cat', 'bark', 'running', 'vehicle', 'joyful'],
    'car': ['engine', 'road', 'puppy', 'computer', 'fast', 'sad'],
    'computer': ['software', 'mouse', 'happy', 'science', 'slow', 'animal'],
    'happy': ['joyful', 'sad', 'running', 'science', 'engine', 'puppy'],
    'science': ['physics', 'biology', 'running', 'software', 'engine', 'cat']
}

# Calculate and organize similarities
similarity_results = {}

for source_word, source_vector in source_words.items():
    similarities = []
    for target_word in target_words[source_word]:
        target_vector = word_embedding_model.loc[word_embedding_model[0] == target_word].to_numpy()[:, 1:][0]
        similarity_score = cosine_similarity(source_vector, target_vector)
        similarities.append((target_word, similarity_score))
    
    # Sorting target words based on similarity score
    similarities.sort(key=lambda x: x[1], reverse=True)
    similarity_results[source_word] = similarities

# Displaying the results
for source_word, similarities in similarity_results.items():
    print(f"Similarity Score for source word: {source_word}")
    print("-" * 19 + "Target Words" +"-" * 19)
    for target_word, similarity_score in similarities:
        seperater = np.round(44 - (len(target_word)))
        print(f"{target_word}" + " " * seperater + f"{similarity_score:.4f}")
    print("\n")
```

```python
# Task A.3
# Nearest neighbors function
def nearest_neighbors(source_vector, target_vectors, k, similarity_metric='cosine'):
    
    if similarity_metric == 'cosine':
        # Calculate cosine similarity for all target vectors
        similarities = np.dot(target_vectors, source_vector) / (np.linalg.norm(target_vectors, axis=1) * np.linalg.norm(source_vector))
    elif similarity_metric == 'dot_product':
        # Calculate dot product similarity for all target vectors
        similarities = np.dot(target_vectors, source_vector)
    else:
        raise ValueError("Invalid similarity metric. Use 'cosine' or 'dot_product'")
   
    # Get indices of k nearest neighbors
    nearest_indices = np.argsort(similarities)[::-1][:k]
    return word_embedding_model.loc[nearest_indices]


# Calculate and display nearest neighbors using both similarity metrics
for source_word, source_vector in source_words.items():
    print(f"Nearest Neighbors for source word: {source_word}")
    
    # Remove the index column before calculating similarity
    word_embedding_vectors = word_embedding_model.iloc[:, 1:].to_numpy()
    
    # nearest_neighbors with cosine and dot product
    print("-" * 5 + " CosineSimilarity " + "-" * 7 + " DotProductSimilarity " + "-" * 5)
    
    cosine_neighbors = nearest_neighbors(source_vector, word_embedding_vectors, k=10, similarity_metric='cosine')
    dot_product_neighbors = nearest_neighbors(source_vector, word_embedding_vectors, k=10, similarity_metric='dot_product')
    
    
    max_length = max(cosine_neighbors[0].apply(len).max(), dot_product_neighbors[0].apply(len).max()) + 5
    
    for idx in range(10):  
        
        cosine_neighbor = cosine_neighbors.iloc[idx]
        cosine_similarity_score = cosine_similarity(source_vector, cosine_neighbor[1:].to_numpy())
        
        dot_neighbor = dot_product_neighbors.iloc[idx]
        dot_similarity_score = cosine_similarity(source_vector, dot_neighbor[1:].to_numpy())
        
        print(f"{cosine_neighbor[0]:<{max_length}} {cosine_similarity_score:.4f}"
              + "   |    " + 
              f"{dot_neighbor[0]:<{max_length}} {dot_similarity_score:.4f}")
    print("\n")
```

**Document embedding as the average of word embeddings (5 points).** Using the word embeddings, the representation of each document is defined as the *mean of the vectors of each document's words*. In particular, given the document $d$, consisting of words $\left[ v_1, v_2, ..., v_{|d|} \right]$, the document representation $\mathbf{e}_d$ is defined as:

$\mathbf{e}_d = \frac{1}{|d|}\sum_{i=1}^{|d|}{\mathbf{e}_{v_i}}$

where $\mathbf{e}_{v}$ is the vector of the word $v$, and $|d|$ is the length of the document.
