```python
# Task A.1
import pandas as pd

# path to txt file
glove_file_path = "./glove.6B/glove.6B.50d.txt"

# creating Dataframe from the pre-trained word embedding
word_embedding_model  = pd.read_csv(glove_file_path, sep=" ", header=None, quoting=3)

# shows some tokens from data 
word_embedding_model[500:510]
```

```python
# Task A.2
# Calculating word-to-word similarities
import numpy as np

# cosine similarity function
def cosine_similarity(vector_a, vector_b):
    
    dot_product = np.dot(vector_a, vector_b)
    
    norm_a = np.linalg.norm(vector_a)
    norm_b = np.linalg.norm(vector_b)
    
    similarity = dot_product / (norm_a * norm_b)
    return similarity

# Source words and their vectors
source_words = {
    'dog': word_embedding_model.loc[word_embedding_model[0] == 'dog'].to_numpy()[:, 1:][0],
    'car': word_embedding_model.loc[word_embedding_model[0] == 'car'].to_numpy()[:, 1:][0],
    'computer': word_embedding_model.loc[word_embedding_model[0] == 'computer'].to_numpy()[:, 1:][0],
    'happy': word_embedding_model.loc[word_embedding_model[0] == 'happy'].to_numpy()[:, 1:][0],
    'science': word_embedding_model.loc[word_embedding_model[0] == 'science'].to_numpy()[:, 1:][0]
}

# Target words for each source word
target_words = {
    'dog': ['puppy', 'cat', 'bark', 'running', 'vehicle', 'joyful'],
    'car': ['engine', 'road', 'puppy', 'computer', 'fast', 'sad'],
    'computer': ['software', 'mouse', 'happy', 'science', 'slow', 'animal'],
    'happy': ['joyful', 'sad', 'running', 'science', 'engine', 'puppy'],
    'science': ['physics', 'biology', 'running', 'software', 'engine', 'cat']
}

# Calculate and organize similarities
similarity_results = {}

for source_word, source_vector in source_words.items():
    similarities = []
    for target_word in target_words[source_word]:
        target_vector = word_embedding_model.loc[word_embedding_model[0] == target_word].to_numpy()[:, 1:][0]
        similarity_score = cosine_similarity(source_vector, target_vector)
        similarities.append((target_word, similarity_score))
    
    # Sorting target words based on similarity score
    similarities.sort(key=lambda x: x[1], reverse=True)
    similarity_results[source_word] = similarities

# Displaying the results
for source_word, similarities in similarity_results.items():
    print(f"Similarity Score for source word: {source_word}")
    print("-" * 5 + "Target Words" + "-" * 5)
    for target_word, similarity_score in similarities:
        seperater = np.round(15 - (len(target_word)))
        print(f"{target_word}" + " " * seperater + f"{similarity_score:.4f}")
    print("\n")
```

```python
# Task A.3
# Nearest neighbors function
def nearest_neighbors(source_vector, target_vectors, k, similarity_metric='cosine'):
    
    if similarity_metric == 'cosine':
        # Calculate cosine similarity for all target vectors
        similarities = np.dot(target_vectors, source_vector) / (np.linalg.norm(target_vectors, axis=1) * np.linalg.norm(source_vector))
    elif similarity_metric == 'dot_product':
        # Calculate dot product similarity for all target vectors
        similarities = np.dot(target_vectors, source_vector)
    else:
        raise ValueError("Invalid similarity metric. Use 'cosine' or 'dot_product'")
   
    # Get indices of k nearest neighbors
    nearest_indices = np.argsort(similarities)[::-1][:k]
    return word_embedding_model.loc[nearest_indices]


# Calculate and display nearest neighbors using both similarity metrics
for source_word, source_vector in source_words.items():
    print(f"Nearest Neighbors for source word: {source_word}")
    
    # Remove the index column before calculating similarity
    word_embedding_vectors = word_embedding_model.iloc[:, 1:].to_numpy()
    
    # nearest_neighbors with cosine and dot product
    print("-" * 5 + " CosineSimilarity " + "-" * 7 + " DotProductSimilarity " + "-" * 5)
    
    cosine_neighbors = nearest_neighbors(source_vector, word_embedding_vectors, k=10, similarity_metric='cosine')
    dot_product_neighbors = nearest_neighbors(source_vector, word_embedding_vectors, k=10, similarity_metric='dot_product')
    
    
    max_length = max(cosine_neighbors[0].apply(len).max(), dot_product_neighbors[0].apply(len).max()) + 5
    
    for idx in range(10):  
        
        cosine_neighbor = cosine_neighbors.iloc[idx]
        cosine_similarity_score = cosine_similarity(source_vector, cosine_neighbor[1:].to_numpy())
        
        dot_neighbor = dot_product_neighbors.iloc[idx]
        dot_similarity_score = cosine_similarity(source_vector, dot_neighbor[1:].to_numpy())
        
        print(f"{cosine_neighbor[0]:<{max_length}} {cosine_similarity_score:.4f}"
              + "   |    " + 
              f"{dot_neighbor[0]:<{max_length}} {dot_similarity_score:.4f}")
    print("\n")
```

<a name="section-taskB"></a><h2 style="color:rgb(0,120,170)">Task B: Document Classification with WE (15 points)</h2>


This task follows the same instruction for document classification as provided in Assignment 1. You are indeed free to reuse any part of your code in Assignment 1 for this task. In Assignment 1, the representation of each document was created using a bag of words representation followed by dimensionality reduction. In this task, the document representations are created from the pre-trained word embeddings.

**Map word embeddings to dictionary words (5 points).** For every word in the dictionary (as discussed and created in Assignment 1), fetch the corresponding word embedding from the pre-trained model. If no embedding is found, initialize the corresponding word embedding randomly.

**Document embedding as the average of word embeddings (5 points).** Using the word embeddings, the representation of each document is defined as the *mean of the vectors of each document's words*. In particular, given the document $d$, consisting of words $\left[ v_1, v_2, ..., v_{|d|} \right]$, the document representation $\mathbf{e}_d$ is defined as:

$\mathbf{e}_d = \frac{1}{|d|}\sum_{i=1}^{|d|}{\mathbf{e}_{v_i}}$

where $\mathbf{e}_{v}$ is the vector of the word $v$, and $|d|$ is the length of the document.

**Classification and evaluation (5 points)** Using these new document representations, apply <ins>three classification algorithms</ins> and report the evaluation results (based on accuracy metric) on the test set.



```python
#Task B.1

# importing dictionary from assignment 1
dictionary_with_threshold = pd.read_json("./dic.json", typ='series').to_dict().values()

def initialize_random_embedding(embedding_dim):
    return np.random.rand(embedding_dim)

embedding_dim = 50
word_embeddings_mapping = {}

for word in dictionary_with_threshold:
    word_embeddings_mapping[word] = word_embedding_model.get(word, initialize_random_embedding(embedding_dim))

df_word_embeddings = pd.DataFrame(list(word_embeddings_mapping.items()), columns=['word', 'embedding'])
df_word_embeddings.head()
```

```python
# reusing code from assignment 1 

from nltk.stem.porter import PorterStemmer
from nltk.stem import WordNetLemmatizer

def stem_words(list_of_tokens: []):
    ps = PorterStemmer()
    
    stem_words = []
    for word in list_of_tokens:
        if word != "<num>" or word != "<dates>":
            stem_words.append(ps.stem(word))
        else:
            stem_words.append(word)
    return stem_words

def lem_words(list_of_tokens: []):
    lemmatizer = WordNetLemmatizer()
    lem_words = []
    for word in list_of_tokens:
        if word != "<num>" or word != "<dates>":
            lem_words.append(lemmatizer.lemmatize(word))
        else:
            lem_words.append(word)
            
    return lem_words

def tokenization(text: pd.Series, normalization_method: str = "lem"):
    corpus = []
    for sentence in text:
        # not using word_tokenize, because i want to have "<num>" and not "<","num", ">"
        corpus.append(sentence.split())
        
    list_of_words = np.concatenate(corpus)
    # gets normalized with one of the methods
    # default method is lem_words
    if normalization_method == "lem":
        return lem_words(list_of_words), corpus
    else:
        return stem_words(list_of_words), corpus


# Task B.2

def document_embedding(text: pd.Series, word_embeddings_mapping, normalization_method: str = "lem"):
    _, corpus = tokenization(text, normalization_method)
    
    document_embeddings = []
    for words in corpus:
        word_embeddings = [word_embeddings_mapping.get(word, np.zeros(embedding_dim)) for word in words]
        document_embedding = np.mean(word_embeddings, axis=0)
        
        document_embeddings.append(document_embedding)
    
    df_document_embeddings = pd.DataFrame(document_embeddings)
    return df_document_embeddings
```

```python
load_train_text = pd.read_csv("nlp2023_24_data/thedeep.subset.train.txt", header=None, names=["Number", "Text", "Label"])
load_validation_text = pd.read_csv("nlp2023_24_data/thedeep.subset.validation.txt", header=None, names=["Number", "Text", "Label"])
load_test_text = pd.read_csv("nlp2023_24_data/thedeep.subset.test.txt", header=None, names=["Number", "Text", "Label"])

df_document_embeddings_train = document_embedding(load_train_text["Text"], word_embeddings_mapping)
df_document_embeddings_test = document_embedding(load_test_text["Text"], word_embeddings_mapping)
df_document_embeddings_validation = document_embedding(load_validation_text["Text"], word_embeddings_mapping)
```

```python
#Task B.3   
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

y_train = load_train_text["Label"]
y_validation = load_validation_text["Label"]
y_test = load_test_text["Label"]

# Logistic Regression
lr = LogisticRegression(max_iter=1000)
lr.fit(df_document_embeddings_train, y_train)
lr_predictions = lr.predict(df_document_embeddings_validation)
lr_accuracy = accuracy_score(y_validation, lr_predictions)
print(f'Logistic Regression Accuracy on Validation Set: {lr_accuracy}')

# Random Forest
rf = RandomForestClassifier()
rf.fit(df_document_embeddings_train, y_train)
rf_predictions = rf.predict(df_document_embeddings_validation)
rf_accuracy = accuracy_score(y_validation, rf_predictions)
print(f'Random Forest Accuracy on Validation Set: {rf_accuracy}')

# Support Vector Machine
svc = SVC()
svc.fit(df_document_embeddings_train, y_train)
svc_predictions = svc.predict(df_document_embeddings_validation)
svc_accuracy = accuracy_score(y_validation, svc_predictions)
print(f'Support Vector Machine Accuracy on Validation Set: {svc_accuracy}')
```

```python
# Logistic Regression
lr_predictions_test = lr.predict(df_document_embeddings_test)
lr_accuracy_test = accuracy_score(y_test, lr_predictions_test)
print(f'Logistic Regression Accuracy on Test Set: {lr_accuracy_test}')

# Random Forest
rf_predictions_test = rf.predict(df_document_embeddings_test)
rf_accuracy_test = accuracy_score(y_test, rf_predictions_test)
print(f'Random Forest Accuracy on Test Set: {rf_accuracy_test}')

# Support Vector Machine
svc_predictions_test = svc.predict(df_document_embeddings_test)
svc_accuracy_test = accuracy_score(y_test, svc_predictions_test)
print(f'Support Vector Machine Accuracy on Test Set: {svc_accuracy_test}')
```
