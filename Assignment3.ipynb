The implementation of the classifier should cover the points below.

**Preprocessing and dictionary (1 point):** Following previous assignments, load the train, validation, and test datasets, apply necessary preprocessing steps, and create a dictionary of words. 

**Data batching (4 points):** Using the dictionary, create batches for any given dataset (train/validation/test). Each batch is a two-dimensional matrix of *batch-size* to *max-document-length*, containing the IDs of the words in the corresponding documents. *Batch-size* and *max-document-length* are two hyper-parameters and can be set to any appropriate values (*Batch-size* must be higher than 1 and *max-document-length* at least 50 words). If a document has more than *max-document-length* words, only the first *max-document-length* words should be kept.

**Word embedding lookup (2 point):** Using `torch.nn.Embedding`, create a lookup for the embeddings of all the words in the dictionary. The lookup is in fact a matrix, which maps the ID of each word to the corresponding word vector. Similar to Assignment 2, use the pre-trained vectors of a word embedding model (like word2vec or GloVe) to initialize the word embeddings of the lookup. Keep in mind that the embeddings of the words in the lookup should be matched with the correct vector in the pretrained word embedding. **Model definition (3 points):** Define the class `ClassificationAverageModel` as a PyTorch model. In the initialization procedure, the model receives the word embedding lookup, and includes it in the model as model's parameters. These embeddings parameters should be trainable, meaning that the word vectors get updated during model training. Feel free to add any other parameters to the model, which might be necessary for accomplishing the functionalities explained in the following.

**Forward function (5 points):** The forward function of the model receives a batch of data, and first fetches the corresponding embeddings of the word IDs in the batch using the lookup. Similar to Assignment 2, the embedding of a document is created by calculating the *element-wise mean* of the embeddings of the document's words. Formally, given the document $d$, consisting of words $\left[ v_1, v_2, ..., v_{|d|} \right]$, the document representation $\mathbf{e}_d$ is defined as:

<center><div>$\mathbf{e}_d = \frac{1}{|d|}\sum_{i=1}^{|d|}{\mathbf{e}_{v_i}}$</div></center>

where $\mathbf{e}_{v}$ is the vector of the word $v$, and $|d|$ is the length of the document. An important point in the implementation of this formula is that the documents in the batch might have different lengths and therefore each document should be divided by its corresponding $|d|$. Finally, this document embedding is utilized to predict the probability of the output classes, done by applying a linear transformation from the embeddings size to the number of classes, followed by Softmax. The linear transformation also belongs to the model's parameters and will be learned in training.

**Loss Function and optimization (2 point):** The loss between the predicted and the actual classes is calculated using Negative Log Likelihood or Cross Entropy. Update the model's parameters using any appropriate optimization mechanism such as Adam.

**Early Stopping (2 points):** After each epoch, evaluate the model on the *validation set* using accuracy.
```python
# Data batching
def batching(data, batch_size=1, max_document_length=50):
    batches = []
    
    # Loop through each batch
    for current_batch in range(batch_size):
        # Calculate the start and end indices for the current batch
        start_index = int(current_batch * (len(data) / batch_size))
        end_index = int((current_batch + 1) * (len(data) / batch_size))
        
        # Extract the batch data within the calculated indices
        batch_data = data[start_index:end_index]
        
        # Truncate each document in the batch to the specified max_document_length
        batch_data = [document[:max_document_length] for document in batch_data]
        
        # Convert words in each document to their corresponding token indices
        batch_id = []
        for document in batch_data:
            new_document = []
            for word in document:
                word_id = token_indices[word]  # Assuming token_indices is a dictionary mapping words to indices
                new_document.append(word_id)
            batch_id.append(new_document)
        
        # Add the processed batch to the list of batches
        batches.append(batch_id)
    
    # Return the list of batches
    return batches

def batching_label(data, num_batches=1):
    batches = []
    
    # Loop through each batch
    for i in range(num_batches):
        # Calculate the start and end indices for the current batch
        start_index = int(i * (len(data) / num_batches))
        end_index = int((i + 1) * (len(data) / num_batches))
        
        # Extract the batch data within the calculated indices
        batch_data = data[start_index:end_index]
        
        # Add the processed batch to the list of batches
        batches.append(batch_data)
    
    # Return the list of batches
    return batches
```

```python
batch_size = 7
max_len_doc = 50
train_batches = batching(data=train_tokens, batch_size=batch_size, max_document_length=max_len_doc)
train_batch_labels = batching_label(data=load_train_text["Label"], num_batches=batch_size)

valid_batches = batching(data=valid_tokens, batch_size=batch_size, max_document_length=max_len_doc)
valid_batch_labels = batching_label(data=load_validation_text["Label"], num_batches=batch_size)
```

```python
def generate_word_vectors(model, token_indices, embedding_dim=300, random_range=(-10, 10)):
    word_vectors = []
    for word in sorted(token_indices.keys(), key=lambda x: token_indices[x]):
        try:
            word_vector = model[word]
        except KeyError:
            word_vector = np.random.uniform(random_range[0], random_range[1], size=(embedding_dim,))
        finally:
            word_vectors.append(word_vector)

    return torch.Tensor(np.array(word_vectors))


# Assuming 'model', 'token_indices', 'embedding_dim', and 'random_range' are defined earlier
word_vectors = generate_word_vectors(model, token_indices, embedding_dim=300, random_range=(-10, 10))
# Create embedding layer from the generated word vectors
embedding_layer = torch.nn.Embedding.from_pretrained(word_vectors)
```

```python
# Model definition / Forward function
class ClassificationAverageModel(nn.Module):
    def __init__(self, embedding, num_classes=12):
        super(ClassificationAverageModel, self).__init__()
        self.embedding_layer = embedding
        self.fc_layer = nn.Linear(self.embedding_layer.embedding_dim, num_classes)

    def forward(self, batch):    
        batch_embeddings = []
        for doc in batch:
            doc_tensor = torch.tensor(doc, dtype=torch.long)  # Convert to tensor
            doc_embeddings = self.embedding_layer(doc_tensor)
            doc_average_embedding = torch.mean(doc_embeddings, dim=0)
            batch_embeddings.append(doc_average_embedding)

        batch_embeddings = torch.stack(batch_embeddings)
        output_scores = self.fc_layer(batch_embeddings)
        output_probabilities = F.softmax(output_scores, dim=1)
        return output_probabilities
```

```python
from torch.utils.tensorboard import SummaryWriter
from torch.optim.lr_scheduler import StepLR

# setting device to gpu
if torch.cuda.is_available():
    device = torch.device("cuda")
    print("GPU is available")
else:
    device = torch.device("cpu")
    print("GPU not available, using CPU")
    
# Setting seed
set_seed = 31
np.random.seed(set_seed)
torch.manual_seed(set_seed)

writer = SummaryWriter()

# setting up model / loss function and optimizer
model = ClassificationAverageModel(embedding)
model = model.to(device)
loss_function = torch.nn.NLLLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
num_epochs = 100
best_accuracy = 0
no_improvement_epochs = 0
early_stopping_threshold = 3

writer = SummaryWriter()  # Initialize the TensorBoard writer

for epoch in tqdm(range(num_epochs)):
    # Training loop
    model.train()
    for i, batch in enumerate(train_batches):
        batch = pad_sequence([torch.tensor(doc, dtype=torch.long) for doc in batch], batch_first=True).to(device)
        
        labels_batch = train_batch_labels[i].tolist()
        
        optimizer.zero_grad()
        outputs = model(batch)
        targets = torch.tensor(labels_batch, dtype=torch.long).to(device)
        loss = loss_function(outputs, targets)
        
        writer.add_scalar("Loss/train", loss, epoch)

        loss.backward()
        optimizer.step()

    # Validation loop
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for i, batch in enumerate(valid_batches):
            batch = pad_sequence([torch.tensor(doc, dtype=torch.long) for doc in batch], batch_first=True).to(device)
            labels_batch = valid_batch_labels[i].tolist()
            
            outputs = model(batch)
            predictions = outputs.argmax(dim=1)
            targets = torch.tensor(labels_batch, dtype=torch.long).to(device)
            correct += (predictions == targets).sum().item()
            total += len(batch)

    accuracy = correct / total
    
    writer.add_scalar("Accuracy/val", accuracy, epoch)
    
    if accuracy > best_accuracy:
        best_accuracy = accuracy
        no_improvement_epochs = 0
        torch.save(model.state_dict(), "ClassificationAverageModel.pt")
    else:
        no_improvement_epochs += 1

    if no_improvement_epochs >= early_stopping_threshold:
        print("Early stopping triggered")
        break
        
    print(f"Epoch: {epoch} | Loss: {loss.item()} | Accuracy: {accuracy}")

# Close the TensorBoard writer
writer.flush()
writer.close()
```

```python
from transformers import BertModel, BertTokenizer

model_name = 'bert-base-uncased'
bert_model = BertModel.from_pretrained(model_name)
BERT_tokenizer = BertTokenizer.from_pretrained(model_name)
```

```python
def tokenize_truncate(tokenizer, texts, max_length=512):
    return tokenizer.batch_encode_plus(texts, max_length=max_length, truncation=True,
                                       padding='max_length', return_tensors='pt')

train_data = tokenize_truncate(BERT_tokenizer, load_train_text["Text"])
test_data = tokenize_truncate(BERT_tokenizer, load_test_text["Text"])
valid_data = tokenize_truncate(BERT_tokenizer, load_validation_text["Text"])
```

```python
class ClassificationBERTModel(nn.Module):
    def __init__(self, bert_model, num_classes=12):
        super(ClassificationBERTModel, self).__init__()
        self.bert = bert_model
        self.fc = nn.Linear(self.bert.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask=None):
        # Ensure attention_mask is not None if provided
        if attention_mask is None:
            attention_mask = torch.ones_like(input_ids)

        # Forward pass through BERT
        outputs = self.bert(input_ids, attention_mask=attention_mask)

        # Extract the [CLS] token representation
        cls_token = outputs.pooler_output

        # Apply softmax to the linear layer output
        output = F.softmax(self.fc(cls_token), dim=1)

        return output
```

```python
from torch.utils.data import DataLoader, TensorDataset

batch_size = 16

# Assuming load_train_text["Label"] and load_validation_text["Label"] contain strings with commas
train_labels = [label for label in load_train_text["Label"]]
valid_labels = [label for label in load_validation_text["Label"]]

train_dataset = TensorDataset(train_data["input_ids"], train_data["attention_mask"], 
                              torch.tensor(train_labels, dtype=torch.long))

val_dataset = TensorDataset(valid_data["input_ids"], valid_data["attention_mask"], 
                            torch.tensor(valid_labels, dtype=torch.long))

# Create DataLoader instances
train_dataloader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_dataloader = DataLoader(val_dataset, batch_size=batch_size)
```

```python
BERT_model = ClassificationBERTModel(bert_model)
BERT_model.to(device)

loss_function = torch.nn.NLLLoss()
optimizer = torch.optim.Adam(BERT_model.parameters(), lr=0.001)
scheduler = StepLR(optimizer, step_size=30, gamma=0.1)
```

```python
num_epochs = 100
best_accuracy = 0
no_improve_epochs = 0
early_stopping_threshold = 3

BERT_model.to(device)

def train_epoch(model, dataloader, optimizer, scheduler, loss_function, device):
    model.train()
    total_loss = 0
    for input_ids, attention_mask, labels in tqdm(dataloader, desc="Training"):
        input_ids, attention_mask, labels = input_ids.to(device), attention_mask.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(input_ids, attention_mask)
        loss = loss_function(outputs, labels)
        loss.backward()
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
    return total_loss / len(dataloader)

def evaluate(model, dataloader, device):
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for input_ids, attention_mask, labels in tqdm(dataloader, desc="Validation"):
            input_ids, attention_mask, labels = input_ids.to(device), attention_mask.to(device), labels.to(device)
            outputs = model(input_ids, attention_mask)
            predictions = outputs.argmax(dim=1)
            correct += (predictions == labels).sum().item()
            total += labels.size(0)
    accuracy = correct / total
    return accuracy

for epoch, _ in tqdm(enumerate(range(num_epochs), 1)):
    train_loss = train_epoch(BERT_model, train_dataloader, optimizer, scheduler, loss_function, device)
    print(f"Epoch: {epoch} | Training Loss: {train_loss}")

    accuracy = evaluate(BERT_model, val_dataloader, device)
    print(f"Epoch: {epoch} | Validation Accuracy: {accuracy}")

    if accuracy > best_accuracy:
        best_accuracy = accuracy
        no_improve_epochs = 0
        torch.save(BERT_model.state_dict(), "BERT_model.pt")
    else:
        no_improve_epochs += 1

    if no_improve_epochs >= early_stopping_threshold:
        print("Early stopping triggered")
        break

```

```python
# Move model to device
BERT_model.to(device)

# Set model to evaluation mode
BERT_model.eval()

# Define a function for evaluation
def evaluate(dataloader, desc):
    correct, total = 0, 0
    with torch.no_grad():
        for batch in tqdm(dataloader, desc=f"{desc}"):
            input_ids, attention_mask, labels = (x.to(device) for x in batch)

            outputs = BERT_model(input_ids, attention_mask)
            predictions = outputs.argmax(dim=1)

            correct += (predictions == labels).sum().item()
            total += labels.size(0)

    accuracy = correct / total
    return accuracy

# Evaluate on validation set
val_accuracy = evaluate(val_dataloader, "Validation")

# Evaluate on test set
test_accuracy = evaluate(test_dataloader, "Testing")
```
